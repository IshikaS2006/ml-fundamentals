```python
def plotting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var],kde=True)

plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)
plt.tight_layout()
```

```python
df['Cholesterol'].value_counts()
```

```python
chmean= df.loc[df['Cholesterol'] != 0, 'Cholesterol'].mean()
restmean= df.loc[df['RestingBP'] != 0, 'RestingBP'].mean()
```

```python
chmean
```

```python
df['Cholesterol']= df['Cholesterol'].replace(0,chmean)
df['Cholesterol']= df['Cholesterol'].round(2)
df['RestingBP']= df['RestingBP'].replace(0,restmean)
df['RestingBP']= df['RestingBP'].round(2)
```

```python
df['Cholesterol'].value_counts()
df['RestingBP'].value_counts()
```
